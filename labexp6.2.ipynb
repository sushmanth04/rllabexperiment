```python
import random

states = ["high", "low"]
alpha = 0.8
beta = 0.6
current_state = "high"
total_reward = 0

print(f"Robot starting state: {current_state}\n")

for step in range(1, 6):
    print(f"--- Step {step} ---")
    print(f"Current Battery: {current_state}")

    if current_state == "high":
        action = "search"
        if random.random() < alpha:
            current_state = "high"
        else:
            current_state = "low"
        reward = 2.0

    elif current_state == "low":
        action = "recharge"
        current_state = "high"
        reward = 0.0

    total_reward += reward
    print(f"Action taken: {action}")
    print(f"Reward received: {reward}")
    print(f"New Battery: {current_state}\n")

print(f"Simulation ended. Total Reward collected: {total_reward}")
```
